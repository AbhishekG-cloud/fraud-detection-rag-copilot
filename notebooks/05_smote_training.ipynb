## DATA Reading

```python
from pathlib import Path

import pandas as pd
from imblearn.over_sampling import SMOTE

raw_data_path = Path("data/creditcard_1.csv")
df = pd.read_csv(raw_data_path)
df = df.sort_values("Time").reset_index(drop = True)

target_column = "Class"

split_index = int(len(df) * 0.8)

train_df = df.iloc[:split_index].copy()
test_df = df.iloc[split_index:].copy()

X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]
```

### First inspect the original state

```python

print("Training shape:", X_train.shape)
print("Test shape:", X_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())

train_distribution = y_train.value_counts(normalize=True).sort_index() * 100
test_distribution = y_test.value_counts(normalize=True).sort_index() * 100

print("\nTraining percentages:")
print(train_distribution)

print("\nTest percentages:")
print(test_distribution)
```

### Test-State Snapshot

```python
test_shape_before = X_test.shape

test_class_counts_before = y_test.value_counts().sort_index().to_dict()

test_class_percentages_before = (
    y_test.value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .to_dict()
)
```

### Apply SMOTE

```python
smote = SMOTE(
    sampling_strategy="auto",
    random_state=42
)

X_train_resampled, y_train_resampled = smote.fit_resample(
    X_train,
    y_train
)
```

### Verify Training Data

```python
print("Original training shape:", X_train.shape)
print("Resampled training shape:", X_train_resampled.shape)

print("\nOriginal training distribution:")
print(y_train.value_counts())

print("\nResampled training distribution:")
print(y_train_resampled.value_counts())
```

### Verify Test Set Was Untouched

```python
test_shape_after = X_test.shape

test_class_counts_after = y_test.value_counts().sort_index().to_dict()

test_class_percentages_after = (
    y_test.value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .to_dict()
)
```

### Integrity Check

```python
assert test_shape_before == test_shape_after
assert test_class_counts_before == test_class_counts_after
assert test_class_percentages_before == test_class_percentages_after

print("Test-set integrity verified.")
```

```python
print("Test shape unchanged:",
      test_shape_before == test_shape_after)

print("Test class counts unchanged:",
      test_class_counts_before == test_class_counts_after)

print("Test class percentages unchanged:",
      test_class_percentages_before == test_class_percentages_after)
```
